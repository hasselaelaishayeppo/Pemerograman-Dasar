```python
# Latihan 1: Menentukan Tahun Kabisat
tahun = int(input("Masukkan Tahun: "))

# Syarat 1: Habis dibagi 4 DAN tidak habis dibagi 100
# Syarat 2: Habis dibagi 400
is_kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

print(f"Apakah Tahun {tahun} Merupakan Tahun Kabisat? {is_kabisat}")

# Latihan 2: Program Mandiri - Kalkulator Persegi Panjang
print("=============================================")
print("   KALKULATOR LUAS DAN KELILING PERSEGI     ")
print("=============================================")

# 1. Fungsi Input (Tipe Float agar bisa desimal)
panjang = float(input("Masukkan Panjang (cm) : "))
lebar   = float(input("Masukkan Lebar (cm)   : "))

# 2. Operasi Aritmetika
luas = panjang * lebar
keliling = 2 * (panjang + lebar)

# 3. Fungsi Output
print("---------------------------------------------")
print(f"Hasil Luas Persegi Panjang     : {luas:.2f} cm²")
print(f"Hasil Keliling Persegi Panjang : {keliling:.2f} cm")
print("=============================================")
```
