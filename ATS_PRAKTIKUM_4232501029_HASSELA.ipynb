```python
# 7. Inisialisasi suhu awal
suhu_saat_ini = 60

# Perulangan untuk menaikkan suhu hingga target 85°C
while suhu_saat_ini < 85:
    suhu_saat_ini += 5

    print(f"Suhu saat ini: {suhu_saat_ini} C")

    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

# Pesan setelah suhu mencapai target
print("Target 85 C tercapai.")
```

```python
# 9. Data pembacaan vibrasi lima unit mesin
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

# Inisialisasi penghitung data tidak valid
jumlah_tidak_valid = 0

# Memeriksa setiap data menggunakan perulangan for
for nomor, nilai in enumerate(data, start=1):
    try:
        vibrasi = float(nilai)

        if vibrasi > 2.0:
            print(f"Warning: Unit {nomor} vibrasi tinggi!")

        if vibrasi > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break

    except ValueError:
        print(f"Unit {nomor}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

# Menampilkan jumlah data tidak valid
print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

```python

# BONUS SOAL PROGRAM MONITORING KONSUMSI ENERGI LISTRIK HARIAN

# Data konsumsi energi listrik selama 5 hari (kWh)
# Data disimpan dalam bentuk teks untuk mengantisipasi kesalahan pembacaan meter
data_energi = ["120.5", "145.0", "ERR", "180.2", "95.3"]

# Inisialisasi variabel
total_energi = 0
jumlah_data_valid = 0
jumlah_data_tidak_valid = 0

print("MONITORING KONSUMSI ENERGI LISTRIK")

# Memeriksa konsumsi energi setiap hari
for hari, data in enumerate(data_energi, start=1):

    try:
        energi = float(data)
        total_energi += energi
        jumlah_data_valid += 1

        print(f"\nHari ke-{hari}: {energi} kWh")

        # Menentukan kondisi konsumsi energi
        if energi > 150:
            print("Status: Konsumsi energi TINGGI!")
        elif energi >= 100:
            print("Status: Konsumsi energi NORMAL.")
        else:
            print("Status: Konsumsi energi RENDAH.")

    except ValueError:
        print(f"\nHari ke-{hari}: Data tidak valid!")
        jumlah_data_tidak_valid += 1
        continue

# Menghitung rata-rata konsumsi energi
if jumlah_data_valid > 0:
    rata_rata = total_energi / jumlah_data_valid

    print("HASIL MONITORING")
    print(f"Total energi: {total_energi:.2f} kWh")
    print(f"Rata-rata energi: {rata_rata:.2f} kWh")
    print(f"Data valid: {jumlah_data_valid}")
    print(f"Data tidak valid: {jumlah_data_tidak_valid}")
else:
    print("Tidak ada data energi yang valid.")
```
