```python
# Input data
jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_buku = int(input("Masukkan harga per buku: Rp "))
diskon_persen = float(input("Masukkan diskon (%): "))
uang_budi = int(input("Masukkan uang Budi: Rp "))
jumlah_orang = int(input("Masukkan jumlah orang: "))

# Perhitungan
total = jumlah_buku * harga_buku
diskon = int(total * diskon_persen / 100)
bayar = total - diskon
kembalian = uang_budi - bayar

# Pembagian iuran secara bilangan bulat
per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang

# Pemeriksaan kecukupan uang
uang_cukup = uang_budi >= bayar

# Menampilkan hasil
print("\nTotal       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", uang_cukup)
```

```python
# Program Soal 2: PLTS Atap Rumah
# Input data
jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel = float(input("Masukkan daya per panel (Wp): "))
jam_penyinaran = float(input("Masukkan penyinaran efektif (jam/hari): "))
efisiensi = float(input("Masukkan efisiensi sistem (contoh 0.8): "))
kebutuhan = float(input("Masukkan kebutuhan listrik (kWh/hari): "))
tarif = float(input("Masukkan tarif listrik (Rp/kWh): "))

# Perhitungan energi
energi_harian = jumlah_panel * daya_panel / 1000 * jam_penyinaran * efisiensi
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif
kekurangan = kebutuhan - energi_harian

# Pemeriksaan kecukupan energi
mencukupi = energi_harian >= kebutuhan

# Menampilkan hasil
print("\nEnergi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kWh):", round(energi_bulanan, 1))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 1))
print("Kekurangan (kWh)    :", round(kekurangan, 1))
print("Mencukupi?          :", mencukupi)
```

```python
# Input data generator
daya_generator = float(input("Masukkan daya generator (kW): "))
jam_operasi = float(input("Masukkan jam operasi per hari: "))
konsumsi_solar = float(input("Masukkan konsumsi solar (liter/kWh): "))
harga_solar = float(input("Masukkan harga solar per liter (Rp): "))
kapasitas_tangki = float(input("Masukkan kapasitas tangki (liter): "))

# Input data kabel
arus = float(input("Masukkan arus kabel (A): "))
hambatan = float(input("Masukkan hambatan kabel (ohm): "))

# Perhitungan energi dan bahan bakar
energi = daya_generator * jam_operasi
solar = energi * konsumsi_solar
biaya = solar * harga_solar

# Perhitungan kapasitas tangki
tangki_cukup = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki % solar

# Perhitungan rugi daya kabel
rugi_kabel = arus ** 2 * hambatan

# Pemeriksaan rugi kabel
rugi_tinggi = rugi_kabel > 150 and arus >= 20
# Menampilkan hasil
print("\nEnergi (kWh)     :", round(energi, 1))
print("Solar (liter)    :", round(solar, 1))
print("Biaya (Rp)       :", round(biaya, 1))
print("Tangki cukup     :", tangki_cukup, "hari")
print("Sisa solar (L)   :", round(sisa_solar, 1))
print("Rugi kabel (W)   :", round(rugi_kabel, 1))
print("Rugi tinggi?     :", rugi_tinggi)
```
